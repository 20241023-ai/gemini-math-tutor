```python
!pip install -q fastapi uvicorn python-multipart google-genai pillow python-dotenv nest_asyncio requests

import os
import requests
from google.colab import userdata

# ==========================================
# 2. 프로젝트 파일 자동 생성 (.env, requirements.txt, main.py, index.html)
# ==========================================

with open('.env', 'w', encoding='utf-8') as f:
    f.write("GEMINI_API_KEY=YOUR_GEMINI_API_KEY_HERE\n")

with open('requirements.txt', 'w', encoding='utf-8') as f:
    f.write("""fastapi
uvicorn
python-multipart
google-genai
pillow
python-dotenv
""")

with open('main.py', 'w', encoding='utf-8') as f:
    f.write('''import io
from fastapi import FastAPI, UploadFile, File, Form, HTTPException
from fastapi.middleware.cors import CORSMiddleware
from google import genai
from google.genai import types
from PIL import Image
from dotenv import load_dotenv

load_dotenv()

app = FastAPI(title="Gemini Math Tutor API")

app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=True,
    allow_methods=["*"],
    allow_headers=["*"],
)

client = genai.Client()

@app.post("/api/solve")
async def solve_math_problem(
    image: UploadFile = File(None),
    text_query: str = Form(None)
):
    contents = []

    if image:
        try:
            image_bytes = await image.read()
            pil_img = Image.open(io.BytesIO(image_bytes))
            contents.append(pil_img)
        except Exception as e:
            raise HTTPException(status_code=400, detail=f"이미지 읽기 실패: {str(e)}")

    if text_query:
        contents.append(text_query)

    if not contents:
        raise HTTPException(status_code=400, detail="이미지나 질문 텍스트를 입력해 주세요.")

    system_instruction = """
    너는 친절하고 명쾌한 수학 튜터야.
    제시된 이미지나 텍스트 문제를 분석하고 다음 구조로 답변해줘:

    1. **인식된 문제**: 이미지에서 인식한 수학 문제를 LaTeX 수식으로 깔끔하게 정리.
    2. **핵심 개념**: 이 문제를 푸는 데 필요한 핵심 수학 개념 요약.
    3. **단계별 풀이**: Step 1, Step 2 등으로 명확하게 나누어 설명.
    4. **함수 그래프식**: 만약 2D 그래프(예: y = x^2 - 4)로 표현할 수 있는 함수식이 문제에 포함되어 있다면, 답변 맨 마지막 줄에 `[GRAPH]: f(x) = ...` 형식으로 한 줄 적어줘. (없다면 생략)
    """

    try:
        response = client.models.generate_content(
            model='gemini-2.5-flash',
            contents=contents,
            config=types.GenerateContentConfig(
                system_instruction=system_instruction,
                temperature=0.2,
            )
        )

        return {
            "status": "success",
            "answer": response.text
        }
    except Exception as e:
        raise HTTPException(status_code=500, detail=f"Gemini API 오류: {str(e)}")

if __name__ == "__main__":
    import uvicorn
    uvicorn.run(app, host="0.0.0.0", port=8000)
''')

with open('index.html', 'w', encoding='utf-8') as f:
    f.write('''<!DOCTYPE html>
<html lang="ko">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>AI 수학 학습 튜터</title>
    <script src="https://cdn.tailwindcss.com"></script>
    <script src="https://cdn.jsdelivr.net/npm/mathjax@3/es5/tex-mml-chtml.js"></script>
    <script src="https://unpkg.com/d3@7/dist/d3.min.js"></script>
    <script src="https://unpkg.com/function-plot/dist/function-plot.js"></script>
    <script src="https://cdn.plot.ly/plotly-2.27.0.min.js"></script>
</head>
<body class="bg-slate-50 text-slate-800 min-h-screen">

    <header class="bg-indigo-600 text-white p-4 shadow-md">
        <div class="max-w-7xl mx-auto flex justify-between items-center">
            <h1 class="text-2xl font-bold flex items-center gap-2">
                <span>📐</span> Gemini AI 수학 학습 튜터
            </h1>
            <span class="text-sm bg-indigo-500 px-3 py-1 rounded-full">GitHub Pages 호스팅</span>
        </div>
    </header>

    <main class="max-w-7xl mx-auto p-4 grid grid-cols-1 lg:grid-cols-12 gap-6 mt-4">

        <section class="lg:col-span-7 flex flex-col gap-4">
            <div class="bg-white p-5 rounded-xl shadow-sm border border-slate-200">
                <h2 class="text-lg font-semibold mb-3 text-slate-700">1. 문제 입력 (이미지 / 텍스트)</h2>
                <form id="solve-form" class="space-y-4">
                    <div>
                        <label class="block text-sm font-medium text-slate-600 mb-1">문제 사진 업로드 (Gemini 이미지 인식)</label>
                        <input type="file" id="image-input" accept="image/*" class="w-full text-sm text-slate-500 file:mr-4 file:py-2 file:px-4 file:rounded-lg file:border-0 file:bg-indigo-50 file:text-indigo-700 hover:file:bg-indigo-100">
                    </div>
                    <div>
                        <label class="block text-sm font-medium text-slate-600 mb-1">질문 직접 입력 / 추가 질문</label>
                        <textarea id="text-input" rows="3" class="w-full p-3 border border-slate-300 rounded-lg focus:ring-2 focus:ring-indigo-500 focus:outline-none" placeholder="예: 이 이차방정식의 근을 구하고 그래프 개형을 설명해줘."></textarea>
                    </div>
                    <button type="submit" id="submit-btn" class="w-full bg-indigo-600 hover:bg-indigo-700 text-white font-medium py-2.5 rounded-lg transition-colors flex justify-center items-center gap-2">
                        🚀 Gemini로 문제 분석 요청
                    </button>
                </form>
            </div>

            <div class="bg-white p-5 rounded-xl shadow-sm border border-slate-200 flex-1 flex flex-col min-h-[350px]">
                <h2 class="text-lg font-semibold mb-3 text-slate-700">2. AI 문제 풀이 및 튜토링</h2>
                <div id="chat-output" class="flex-1 overflow-y-auto p-4 bg-slate-50 rounded-lg border border-slate-100 text-sm leading-relaxed">
                    <p class="text-slate-400 text-center mt-10">사진을 첨부하거나 질문을 입력하면 Gemini AI가 풀어줍니다.</p>
                </div>
            </div>
        </section>

        <section class="lg:col-span-5 flex flex-col gap-6">
            <div class="bg-white p-5 rounded-xl shadow-sm border border-slate-200">
                <h2 class="text-lg font-semibold mb-2 text-slate-700">📊 수학 수식/함수 시각화</h2>
                <p class="text-xs text-slate-500 mb-3">문제 내 함수식이 감지되면 그래프가 자동으로 그려집니다.</p>
                <div id="function-graph" class="w-full h-64 bg-slate-50 rounded-lg border border-slate-200 flex justify-center items-center"></div>
            </div>

            <div class="bg-white p-5 rounded-xl shadow-sm border border-slate-200">
                <h2 class="text-lg font-semibold mb-2 text-slate-700">📈 성취도 분석 (Plotly)</h2>
                <div id="radar-chart" class="w-full h-60"></div>
            </div>
        </section>

    </main>

    <script>
        const BACKEND_URL = "http://localhost:8000";

        function initDashboard() {
            const radarData = [{
                type: 'scatterpolar',
                r: [85, 70, 95, 60, 75],
                theta: ['다항식', '방정식', '수열', '미적분', '확률과통계'],
                fill: 'toself',
                fillcolor: 'rgba(99, 102, 241, 0.2)',
                line: { color: '#6366f1' }
            }];
            const radarLayout = {
                polar: { radialaxis: { visible: true, range: [0, 100] } },
                margin: { t: 20, b: 20, l: 30, r: 30 },
                showlegend: false
            };
            Plotly.newPlot('radar-chart', radarData, radarLayout, {responsive: true});
        }

        function drawFunction(fnString) {
            try {
                functionPlot({
                    target: '#function-graph',
                    width: 380,
                    height: 240,
                    grid: true,
                    data: [{ fn: fnString, color: '#4f46e5' }]
                });
            } catch (e) {
                console.log("그래프 생성 중 오류:", e);
            }
        }

        drawFunction("x^2 - 4");
        initDashboard();

        document.getElementById('solve-form').addEventListener('submit', async (e) => {
            e.preventDefault();
            const submitBtn = document.getElementById('submit-btn');
            const chatOutput = document.getElementById('chat-output');
            const imageInput = document.getElementById('image-input').files[0];
            const textInput = document.getElementById('text-input').value;

            if (!imageInput && !textInput.trim()) {
                alert("사진을 첨부하거나 질문을 입력해 주세요.");
                return;
            }

            submitBtn.disabled = true;
            submitBtn.innerText = "⏳ Gemini가 문제를 분석 중입니다...";

            const formData = new FormData();
            if (imageInput) formData.append('image', imageInput);
            if (textInput) formData.append('text_query', textInput);

            try {
                const response = await fetch(`${BACKEND_URL}/api/solve`, {
                    method: 'POST',
                    body: formData
                });

                const data = await response.json();

                if (data.status === "success") {
                    let answerText = data.answer;

                    if (answerText.includes("[GRAPH]:")) {
                        const graphPart = answerText.split("[GRAPH]:")[1].trim();
                        const rawFn = graphPart.replace("f(x) =", "").trim();
                        drawFunction(rawFn);
                        answerText = answerText.split("[GRAPH]:")[0];
                    }

                    chatOutput.innerHTML = `
                        <div class="prose max-w-none text-slate-700 whitespace-pre-line">${answerText}</div>
                    `;

                    MathJax.typesetPromise([chatOutput]);
                } else {
                    alert("오류: " + data.detail);
                }
            } catch (err) {
                alert("서버 통신 실패! 백엔드가 실행 중인지 확인하세요.");
                console.error(err);
            } finally {
                submitBtn.disabled = false;
                submitBtn.innerText = "🚀 Gemini로 문제 분석 요청";
            }
        });
    </script>
</body>
</html>
''')

print("📂 로컬 파일 생성 완료!")

# ==========================================
# 3. GitHub 자동 리포지토리 생성 및 Push
# ==========================================
print("\n--- 🐙 GitHub 인증 정보 입력 ---")

try:
    github_token = userdata.get('GITHUB_TOKEN')
    github_user = userdata.get('GITHUB_USER')
    repo_name = userdata.get('GITHUB_REPO')
except:
    github_token = None
    github_user = None
    repo_name = None

if not github_token:
    github_token = input("GitHub Personal Access Token (PAT): ").strip()
if not github_user:
    github_user = input("GitHub Username: ").strip()
if not repo_name:
    repo_name = input("GitHub 저장소 이름 (예: math-tutor): ").strip()

# GitHub API로 원격 저장소가 없으면 자동 생성
headers = {
    "Authorization": f"token {github_token}",
    "Accept": "application/vnd.github.v3+json"
}
res = requests.post("https://api.github.com/user/repos", json={"name": repo_name, "auto_init": False}, headers=headers)

if res.status_code == 201:
    print(f"✨ GitHub 저장소 '{repo_name}'가 새롭게 생성되었습니다.")
elif res.status_code == 422:
    print(f"ℹ️ 기존 저장소 '{repo_name}'에 연결합니다.")

# Git 초기화 및 main 브랜치 푸시
!rm -rf .git
!git init -b main
!git config --global user.email "{github_user}@users.noreply.github.com"
!git config --global user.name "{github_user}"

!git add requirements.txt main.py index.html .env
!git commit -m "Deploy math tutor app via Colab"

remote_url = f"https://{github_token}@github.com/{github_user}/{repo_name}.git"
!git remote add origin {remote_url}
!git push -u origin main --force

print(f"\n✅ 완료되었습니다! GitHub 저장소 주소:")
print(f"🔗 https://github.com/{github_user}/{repo_name}")
print("\n📌 이제 저장소 Settings > Pages 이동 시 'main' 브랜치가 선택 가능합니다.")
```
